# Data Loading from AWS S3 Demo

This notebook shows how to load data from an AWS S3 bucket with SQL/PySpark and how to save data from S3 into Databricks tables.

### Prerequisites
- Connect to your S3 bucket, by creating a storage credential and an external location for S3 from the Databricks Catalog Explorer (follow the steps from [here](https://docs.databricks.com/aws/en/connect/unity-catalog/cloud-storage/s3/s3-external-location-manual)).
- Upload the necessary CSV files (from `sample-data/cafe_sales`) to your S3 bucket and modify the `base_file_path` accordingly.

```python
base_file_path = "s3://szk-databricks-training-demo-s3/cafe_sales"
```

### Read all .csv files from the S3 bucket with SQL

```python
%sql
-- Modify the "s3://szk-databricks-training-demo-s3" path to your own S3 bucket
SELECT * FROM read_files('s3://szk-databricks-training-demo-s3/cafe_sales/*.csv')
```

### Read all .csv files from the S3 bucket with python

```python
df = (spark.read
  .format("csv")
  .option("header", "true")
  .option("inferSchema", "true")
  .load(base_file_path)
)

df.display()
```

### Create table with the data from the .csv files from the S3 bucket

```python
%sql
-- Modify the "s3://szk-databricks-training-demo-s3" path to your own S3 bucket

CREATE TABLE IF NOT EXISTS databricks_training.demo.cafe_sales
AS SELECT *
FROM read_files('s3://szk-databricks-training-demo-s3/cafe_sales/*.csv')
```

### Display the data from the table

```python
%sql
SELECT * FROM databricks_training.demo.cafe_sales
```
